# **CHAINS In Langchain**

```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv(override=True)

api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
    raise ValueError("API Key not found in .env")

client = genai.Client(api_key=api_key)

from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0)
```

# **First Chain**

```python
# TASK -1 [Prompt]

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant"),
    ("human", "{input}")
])
```

```python
# TASK - 2 [LLM]

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash",temperature=0)
```

```python
# TASK - 3 [Str Parser]

str_parser = StrOutputParser()
```

### **Manual Invocation**


```python
ready_prompt= prompt_template.invoke({"input": "What is the capital of France?"})

res = llm.invoke(ready_prompt)

final_result = str_parser.invoke(res)
```

```python
from rich.console import Console
from rich.panel import Panel

console = Console()

ready_prompt = prompt_template.invoke({
    "input": "What is the capital of France?"
})

res = llm.invoke(ready_prompt)

final_result = str_parser.invoke(res)

console.print(
    Panel(
        final_result,
        title="[bold cyan]Gemini Response[/bold cyan]",
        border_style="cyan",
        padding=(1, 2)
    )
)
```

### **Chain Invocation**


```python
chain = prompt_template | llm | str_parser

chain.invoke({"input": "What is the capital of France?"})
```

```python
from rich.console import Console
from rich.panel import Panel

console = Console()

# 1. Build the chain
chain = prompt_template | llm | str_parser

# 2. Run the chain
final_result = chain.invoke({
    "input": "What is the capital of France?"
})

# 3. Display the result
console.print(
    Panel(
        final_result,
        title="[bold cyan]Gemini Response[/bold cyan]",
        border_style="cyan",
        padding=(1, 2)
    )
)
```

```python
from langchain_core.runnables import RunnableSequence

chain_1 = RunnableSequence(prompt_template, llm, str_parser)

chain_1.invoke("What is the capital of France?")
```
