```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv(override=True)

api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
    raise ValueError("API Key not found in .env")

client = genai.Client(api_key=api_key)

from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0)
```

# **CHAIN WITH CUSTOM RUNNABLE**

```python
# TASK -1 [Prompt]

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant"),
    ("human", "{input}")])
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0)
```

```python
# TASK - 3 [Str Parser]

str_parser = StrOutputParser()
```

```python
# TASK - 4 [Custom Runnable]
from langchain_core.runnables import RunnableLambda

def dictionary_maker(text:str)-> dict:

    return {"text" : text}

dictionary_maker_runnable = RunnableLambda(dictionary_maker)
```

```python
test_result = dictionary_maker("Hello world")

print(test_result)
```

```python
# TASK - 5 [Template For Post]

prompt_post = ChatPromptTemplate.from_messages(
    messages=[
        ("system", "You're a social media post generator. "),
        ("human", "Create a post for the following text for LinkedIn: {text}"),
    ]
)
```

```python
# TASK - 6 [Generate the Post]
llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0)
```

```python
# TASK - 7 [Str Parser]

str_parser = StrOutputParser()
```

```python
# CHAIN

chain = prompt_template | llm | str_parser | dictionary_maker_runnable | prompt_post | llm | str_parser

chain.invoke("What is the capital of France?")
```

```python
from rich.console import Console
from rich.panel import Panel

console = Console()
```

```python
final_result = chain.invoke({
    "input": "What is the capital of France?"
})

# 3. Display the result
console.print(
    Panel(
        final_result,
        title="[bold cyan]Gemini Response[/bold cyan]",
        border_style="cyan",
        padding=(1, 2)
    )
)
```
