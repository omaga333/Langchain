```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv(override=True)

api_key = os.getenv("GEMINI_API_KEY2")

if not api_key:
    raise ValueError("API Key not found in .env")

client = genai.Client(api_key=api_key)

response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="Bro, tell me a fun fact"
)

print(response.text)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0
)

my_msg = [
    SystemMessage(content="You are a rude assistant."),
    HumanMessage(content="Tell me what is the output of 3 + 2?")
]

response = llm.invoke(my_msg)

print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate

dynamic_prompt = PromptTemplate.from_template(
    "Write a fun fact about {topic}"
)
```

```python
print(dynamic_prompt.input_variables)
```

```python
user_input = input("Enter a topic for fun fact: ")

ready_prompt = dynamic_prompt.invoke({
    "topic": user_input
})
```

```python
print(ready_prompt)
print(ready_prompt.text)
print(ready_prompt.to_string())
```

```python
response = llm.invoke(ready_prompt)
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

ChatPromptTemplate = ChatPromptTemplate.from_messages([

     ("system", "You are a polite assistant."),
    ("human", "Write a fun fact about {topic}.")
]
)
```

```python
user_input = input("Enter a topic for fun fact: ")

ready_prompt = ChatPromptTemplate.invoke({
    "topic": user_input
})

response = llm.invoke(ready_prompt)

print(response.content)
```

```python
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an assistant. Your tone is {tone}."),
    ("human", "Write a fun fact about {topic}.")
])
```

```python
topic = input("Enter a topic: ")
tone = input("Enter a tone: ")
```

```python
ready_prompt = chat_prompt.invoke({
    "topic": topic,
    "tone": tone
})
```

```python
response = llm.invoke(ready_prompt)

print(response.content)
```
